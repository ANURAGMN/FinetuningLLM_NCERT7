```python
import gradio as gr
from langchain_community.llms import HuggingFaceEndpoint
from langchain.chains import LLMChain
from langchain_core.prompts import PromptTemplate

template = """Below describes conversation happening between teacher and student.
Predict what the next statement from Teacher should be.


Question from student: {question}
"""

prompt = PromptTemplate.from_template(template)
repo_id = "markkevins109/finetuned_unsloth_llama-3-8b-bnb-4bit_on_TSConvo"

llm = HuggingFaceEndpoint(
    repo_id=repo_id,
    max_length=128,
    temperature=0.9,
    huggingfacehub_api_token='',
    task="text-generation",
)

llm_chain = prompt | llm

def generate_response(question):
    return llm_chain.invoke({"question": question})

iface = gr.Interface(
    fn=generate_response,
    inputs=gr.Textbox(lines=4, placeholder="Enter the conversation here..."),
    outputs="text",
    title="Teacher-Student Conversation Prediction",
    description="Enter a conversation between a teacher and a student, and the model will predict the next statement from the teacher."
)

iface.launch()
```

```python
import gradio as gr
from langchain_community.llms import HuggingFaceEndpoint
from langchain.chains import LLMChain
from langchain_core.prompts import PromptTemplate

template = """Below describes conversation happening between teacher and student.
Predict what the next statement from Teacher should be.


Question from student: {question}
"""

prompt = PromptTemplate.from_template(template)

model_options = {
    "LLaMA-3 8B": "markkevins109/finetuned_unsloth_llama-3-8b-bnb-4bit_on_TSConvo",
    "Gemma 7B": "markkevins109/fine_tuned_unsloth_gemma-7b-bnb-4bit_on_TSConvo"
}

def generate_response(question, model_choice):
    repo_id = model_options[model_choice]
    llm = HuggingFaceEndpoint(
        repo_id=repo_id,
        max_length=128,
        temperature=0.9,
        huggingfacehub_api_token='',
        task="text-generation",
    )
    llm_chain = prompt | llm
    return llm_chain.invoke({"question": question})

iface = gr.Interface(
    fn=generate_response,
    inputs=[
        gr.Textbox(lines=4, placeholder="Enter the conversation here..."),
        gr.Radio(choices=list(model_options.keys()), label="Choose Model")
    ],
    outputs="text",
    title="Teacher-Student Conversation Prediction",
    description="Enter a conversation between a teacher and a student, select a model, and the model will predict the next statement from the teacher."
)

iface.launch()
```
