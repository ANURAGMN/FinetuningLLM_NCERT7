```python
import PyPDF2
from groq import Groq

def extract_text_from_pdf(pdf_path):
    with open(pdf_path, "rb") as file:
        reader = PyPDF2.PdfReader(file)
        text = ""
        for page in reader.pages:
            text += page.extract_text()
        return text

def structure_text_with_groq(raw_text):
    client = Groq(api_key="")  # Initialize Groq client with API key
    completion = client.chat.completions.create(
        model="llama-3.3-70b-versatile",  # Groq model
        messages=[
            {"role": "system", "content": "You are a helpful assistant."},
            {
                "role": "user",
                "content": (
                    "Take the following text and structure it by identifying all topics. "
                    "For each topic, add a prefix '### ' followed by the topic name. The content "
                    "under each topic should remain as paragraphs. Ensure the structure is clean and readable."
                    "get the content only from this text \n\n"
                    f"{raw_text}"
                ),
            },
        ],
        temperature=1,
        #max_completion_tokens=32768,
        top_p=1,
        stream=True,
        stop=None,
    )
    structured_text = ""
    for chunk in completion:
        structured_text += chunk.choices[0].delta.content or ""
    return structured_text

pdf_path = r"C:\Users\mark kevin\Downloads\gesc1dd (1)\gesc111.pdf"
output_file = "structured_topics_using_GROQfull11.txt"

# Step 1: Extract text from the PDF
raw_text = extract_text_from_pdf(pdf_path)
print("The first few lines of the text extracted from the pdf is:")
print("\n".join(raw_text.splitlines()[:10]))


# Step 2: Structure the text using Groq API
structured_text = structure_text_with_groq(raw_text)

# Step 3: Save the structured text to a file
with open(output_file, "w", encoding="utf-8") as file:
    file.write(structured_text)

print(f"Structured text saved to '{output_file}'.")
```

```python
import re
import csv

# File paths
input_file = "structured_topics_using_GROQ2.txt"
output_file = "topics_and_content.csv"

# Read the input text file
with open(input_file, "r", encoding="utf-8") as file:
    content = file.read()

# Use regex to split the topics and their content
pattern = r"###\s*(.+?)\n(.*?)(?=(?:\n###|\Z))"
matches = re.findall(pattern, content, re.S)

# Write to CSV
with open(output_file, "w", newline="", encoding="utf-8") as csvfile:
    writer = csv.writer(csvfile)
    writer.writerow(["Topic", "Content"])
    for topic, text in matches:
        writer.writerow([topic.strip(), text.strip()])

print(f"CSV file saved as '{output_file}'.")
```
