```python
import pandas as pd
from langchain_community.llms import HuggingFaceEndpoint
from langchain.chains import LLMChain
from langchain_core.prompts import PromptTemplate

# Define the model and prompt
template = """Below describes conversation happening between teacher and student.
Predict what the next statement from Teacher should be.
Now You are a friendly teacher providing clear, concise answers to student questions. Keep responses brief but engaging.
The give Input consists of a conversation between the teacher and the student now your role is to predict what the next statement from Teacher should be.

Use this as an example
Input: Teacher:  Hey, Rohan! How's it going? Did you watch the cricket match last night? Teacher:  That's great! You know, just like how a cricket team needs a good strategy to win a match, a Wastewater Treatment Plant (WWTP) needs a good strategy to treat wastewater. Have you ever thought about what happens to the water that goes down the drain after you use it? Teacher:  That's right! It does go to the sewer, and from there, it goes to a WWTP. But what do you think happens to the water at the treatment plant? Teacher:  Exactly! They do clean it, but it's not just a simple process. The treatment plant uses physical, chemical, and biological processes to remove all the contaminants from the water. Can you think of what kind of contaminants might be in the water? Teacher:  That's right! Dirt, trash, and other physical contaminants. But there are also chemical contaminants like pollutants from factories and biological contaminants like bacteria and viruses. The treatment plant needs to remove all of these to make the water safe for release into the environment.
The output that you generate should be something like what is below
Output:That's right! Dirt, trash, and other physical contaminants. But there are also chemical contaminants like pollutants from factories and biological contaminants like bacteria and viruses. The treatment plant needs to remove all of these to make the water safe for release into the environment.

Question from student: {question}
"""

prompt = PromptTemplate.from_template(template)

repo_id = "markkevins109/fine_tuned_unsloth_gemma-7b-bnb-4bit_on_TSConvo"

llm = HuggingFaceEndpoint(
    repo_id=repo_id,
    max_length=128,
    temperature=0.9,
    huggingfacehub_api_token='hf',
    task="text-generation",
)

llm_chain = prompt | llm

# Load CSV file
csv_file_path = '/content/Data_for_validation.csv'  # Change this to your actual file path
df = pd.read_csv(csv_file_path)

# Check if 'Input' column exists in the CSV
if 'Input' not in df.columns:
    raise ValueError("CSV file must have an 'Input' column.")

# Function to generate response for each question
def generate_teacher_response(question):
    response = llm_chain.invoke({"question": question})
    return response  # Return the string directly, no need to access 'generated_text'

# Generate content and append it to a new column in the DataFrame
df['Generated_Response'] = df['Input'].apply(generate_teacher_response)

# Save the updated DataFrame to a new CSV file
output_file_path = 'updated_conversations.csv'  # Path to save the updated CSV
df.to_csv(output_file_path, index=False)

print(f"Updated CSV saved at: {output_file_path}")
```

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Load the CSV file
df = pd.read_csv("/content/updated_conversations12.csv")

# Define the column names
col1 = "Output"  # Replace with actual column name
col2 = "Generated_Response"  # Replace with actual column name

# Convert text into TF-IDF vectors
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(df[col1].astype(str).tolist() + df[col2].astype(str).tolist())

# Compute cosine similarity
cosine_sim = cosine_similarity(tfidf_matrix[:len(df)], tfidf_matrix[len(df):])

# Store results in a new column
df["cosine_similarity"] = [cosine_sim[i][i] for i in range(len(df))]

# Calculate the average cosine similarity
average_similarity = df["cosine_similarity"].mean()

# Save the updated CSV
df.to_csv("output_with_similarity_llama.csv", index=False)

# Display results
print(df[["cosine_similarity"]])
print(f"Overall Average Cosine Similarity: {average_similarity:.4f}")
```
